```python
import pandas as pd
import geopandas as gpd
import os
import glob
```

```python
df_census = pd.read_csv('census.csv', index_col=0)
df_census
```

```python
df_census.rename(columns={'years':'year', 'geography':'state_name'}, inplace=True)
df_census.head(5)
```

```python
gdf_shape = gpd.read_file('../../geo_boundaries/CAN/lpr_000b21a_e.shp')
gdf_shape
```

```python
df_census = df_census.merge(gdf_shape[['PRENAME','PREABBR']], how='left', left_on='state_name', right_on='PRENAME').drop(columns=['PRENAME']).rename(columns={'PREABBR':'state_code'})
```

```python
df_census.shape
# 130 rows × 209 columns
```

```python
cols_rename = ['total_births',
                'live_births',
                'fetal_deaths_(stillbirths)',
                'perinatal_deaths_by_year',
                '13_to_19_years',
                '20_to_34_years',
                '35_to_49_years',
                '50_years_and_above',
                '12_to_17_years',
                '18_to_34_years',
                '50_to_64_years',
                '65_years_and_above',
                '12_to_17_years_male',
                '18_to_34_years_male',
                '35_to_49_years_male',
                '50_to_64_years_male',
                '65_years_and_above_male',
                '12_to_17_years_female',
                '18_to_34_years_female',
                '35_to_49_years_female',
                '50_to_64_years_female',
                '65_years_and_above_female',
                'pop_females_heavy_drinking',
                'pop_females_heavy_drinking_12_to_17_years',
                'pop_females_heavy_drinking_18_to_34_years',
                'pop_females_heavy_drinking_35_to_49_years',
                'pop_females_heavy_drinking_50_to_64_years_',
                'pop_females_heavy_drinking_65_years_and_over',
                'pop_male_heavy_drinking',
                'pop_heavy_drinking_males_12_to_17_years',
                'pop_heavy_drinking_males_18_to_34_years',
                'pop_heavy_drinking_males_35_to_49_years',
                'pop_heavy_drinking_males_50_to_64_years',
                'pop_heavy_drinking_males_65_years_and_over',
                'pop_heavy_drinking',
                'pop_heavy_drinking_12_to_17_years',
                'pop_heavy_drinking_18_to_34_years',
                'pop_heavy_drinking_35_to_49_years',
                'pop_heavy_drinking_50_to_64_years',
                'pop_heavy_drinking_65_years_and_over'
              ]
obj = {}
for c in cols_rename:
    for k in ['births','deaths','years','heavy_drinking']:
        if k in c:
            obj[c] = f"{k}_{c.replace(f'_{k}','')}"

df_census.rename(columns=obj, inplace=True)
df_census
```

```python
fcols = ['state_name','state_code','year']
lcols = sorted(list(set(df_census.columns) - set(fcols)))

df_census = df_census[fcols + lcols]
df_census
```

```python
FN = 'CAN_all_census_stats.csv'
df_census.to_csv(FN)
```
