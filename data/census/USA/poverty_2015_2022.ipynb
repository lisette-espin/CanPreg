```python
import pandas as pd

# Path to the Excel file (replace with the actual file path)
file_path = ios.get_files_from_pattern(ios.path_join(ROOT,"*pov*_mod.xlsx"))[0]

# Read the Excel file with all sheets
excel_data = pd.ExcelFile(file_path)

# Initialize a list to collect the data
data = pd.DataFrame()

# Process each sheet (tab)
for sheet_name in excel_data.sheet_names:
    # Read the data from the current sheet
    tmp = excel_data.parse(sheet_name)
    
    # Identify the year from the sheet name
    try:
        year = int(sheet_name)
    except:
        continue
    
    print(year)
    tmp.drop(index=[0,1,2], inplace=True)    
    tmp.reset_index(inplace=True, drop=True)
    tmp.columns = tmp.loc[0]  # Set the header row
    tmp.rename(columns={'State':'state_name', 'Number in poverty':f'{metric}_total'}, inplace=True)
    tmp = tmp[1:]
    tmp.drop(columns=[c for c in tmp.columns if c not in ['state_name',f'{metric}_total']], inplace=True)
    tmp.loc[:,'year'] = year
    tmp.reset_index(inplace=True, drop=True)
    tmp.columns.name = None

    
    tmp.set_index(["state_name",'year'], inplace=True)
    data = pd.concat([data, tmp], ignore_index=False)
    
data.shape
```
