```python
import geopandas as gpd
```

```python
files = ios.get_files_from_pattern("*2015_2022.csv")
files
```

```python
df_census = None
for fn in files:
    df = ios.read_csv(fn, index_col=['state_name','year'])
    df_census = df_census.join(df) if df_census is not None else df.copy()
df_census.shape
```

```python
df_census
```

```python
gdf_shape = gpd.read_file('../../geo_boundaries/USA/us-state-boundaries.zip')
gdf_shape.head(5)
```

```python
df_census = df_census.reset_index()
df_census = df_census.merge(gdf_shape[['basename','stusab']], how='left', left_on='state_name', right_on='basename').drop(columns=['basename']).rename(columns={'stusab':'state_code'})
df_census
```

```python
df_census.head(5)
```

```python
fcols = ['state_name','state_code','year']
lcols = sorted(list(set(df_census.columns) - set(fcols)))

df_census = df_census[fcols + lcols]
df_census
```

```python
FN_FINAL = 'USA_all_census_stats.csv'
ios.save_csv(df_census.reset_index(), FN_FINAL)
```
