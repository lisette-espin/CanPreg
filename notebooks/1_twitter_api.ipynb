```python
import requests
import os
import json
```

```python
import sys
sys.path.append('../code/libs')
import ios
```

```python
bearer_token = ios.read_lines_as_list('../auth/bearer_token')[0]
```

```python
search_url = "https://api.twitter.com/2/tweets/search/all"

# Optional params: start_time,end_time,since_id,until_id,max_results,next_token,
# expansions,tweet.fields,media.fields,poll.fields,place.fields,user.fields
query_params = {"query":"(pregnancy cannabis OR pregnancy weed OR pregnancy pot) -is:retweet place_country:US lang:en has:geo",
                "start_time":"2012-01-01T00:00:00Z",
                "tweet.fields":"author_id,attachments,conversation_id,created_at,entities,geo,in_reply_to_user_id,lang,public_metrics",
                "user.fields":"created_at,description,entities,id,location,name,profile_image_url,verified,username,pinned_tweet_id"
                }

def bearer_oauth(r):
    """
    Method required by bearer token authentication.
    """

    r.headers["Authorization"] = f"Bearer {bearer_token}"
    r.headers["User-Agent"] = "v2FullArchiveSearchPython"
    return r


def connect_to_endpoint(url, params):
    response = requests.request("GET", search_url, auth=bearer_oauth, params=params)
    print(response.status_code)
    if response.status_code != 200:
        raise Exception(response.status_code, response.text)
    return response.json()


def main():
    json_response = connect_to_endpoint(search_url, query_params)
    print(json.dumps(json_response, indent=4, sort_keys=True))


if __name__ == "__main__":
    main()
```
